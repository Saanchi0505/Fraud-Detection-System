# Fraud Detection – Model Training

Objective:
- Train multiple ML models
- Handle class imbalance
- Compare model performance
- Select best base model

```python
import joblib
import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from xgboost import XGBClassifier
```

```python
X_train = joblib.load("../data/X_train_processed.pkl")
X_test = joblib.load("../data/X_test_processed.pkl")
y_train = joblib.load("../data/y_train.pkl")
y_test = joblib.load("../data/y_test.pkl")
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)
```

# Logistic Regression

```python
log_model = LogisticRegression(class_weight='balanced',max_iter=1000)
log_model.fit(X_train, y_train)
log_preds = log_model.predict(X_test)
print("Logistic Regression Classification Report:")
print(classification_report(y_test, log_preds))
```

### Logistic Regression Insight

- Handles imbalance using class_weight.
- Provides good baseline performance.
- Linear model → may not capture complex fraud patterns.

## Random Forest Model

```python
rf_model = RandomForestClassifier(n_estimators=200,class_weight='balanced',random_state=42)
rf_model.fit(X_train, y_train)
rf_preds = rf_model.predict(X_test)
print("Random Forest Classification Report:")
print(classification_report(y_test, rf_preds))
```

### Random Forest Insight

- Better recall and precision than Logistic Regression.
- Captures nonlinear relationships.
- Tree-based probability estimates may saturate at high values.

## XGBoost 

```python
fraud = np.sum(y_train == 1)
non_fraud = np.sum(y_train == 0)

scale_pos_weight = non_fraud/fraud
print(f"Scale Pos Weight: {scale_pos_weight:.2f}")
```

```python
xgb_model = XGBClassifier(n_estimators = 200,scale_pos_weight=scale_pos_weight, eval_metric='logloss', learning_rate=0.1,max_depth = 6,random_state=42)
xgb_model.fit(X_train, y_train)
xgb_preds = xgb_model.predict(X_test)
print("XGBoost Classification Report:")
print(classification_report(y_test, xgb_preds))
```
